```python
import pydicom, glob, os

root = '/content/drive/MyDrive/COMPUTER SCIENCE/NAM4_KH1_(2026-2027)/CT208E/SOURCES/DATASET/LIDC-IDRI-0001'

seg_files = sorted(glob.glob(f"{root}/SEG*/*.dcm"))
print("Số file SEG:", len(seg_files))

for p in seg_files:
    ds = pydicom.dcmread(p)
    print("\n---", os.path.basename(os.path.dirname(p)))
    print("frames:", ds.NumberOfFrames)
    print("SeriesDescription:", ds.get("SeriesDescription", ""))
    print("ContentCreatorName:", ds.get("ContentCreatorName", ""))
    print("ContentDescription:", ds.get("ContentDescription", ""))
    for s in ds.SegmentSequence:
        print("segment:", s.SegmentNumber, "|", s.SegmentLabel,
              "|", s.get("SegmentAlgorithmName", ""),
              "|", s.get("SegmentAlgorithmType", ""))
    pf = ds.PerFrameFunctionalGroupsSequence[0]
    print("có DerivationImageSequence:", "DerivationImageSequence" in pf)
    print("có PlanePositionSequence:", "PlanePositionSequence" in pf)
```
